## Load and Clean ACS data

```python
def load_acs_tables(filepath,cols,col_names):
    df = pd.read_csv(filepath)
    
    df["GEO_ID"] = df["GEO_ID"].astype(str).str.replace('1400000US', '', regex=False).str.zfill(11)  
    
    df = df.drop(df.index[0])#Drop second row
    df = df[cols]
    
    df = df.rename(columns=col_names)
    
    for col in df.columns:
        if col != 'geo_id':
            df[col] = pd.to_numeric(df[col], errors='coerce')
    
    return df 
```

```python
# Create ACS dataframes
age_df = load_acs_tables('data/ACS/S0101_AgeSex.csv',s0101_cols,s0101_colnames)

hh_family_df = load_acs_tables('data/ACS/S1101_HHFamilies.csv',s1101_cols,s1101_colnames)

poverty_df = load_acs_tables('data/ACS/S1701_PovertyStatus.csv',s1701_cols,s1701_colnames) 

disability_df = load_acs_tables('data/ACS/S1810_Disability.csv',s1810_cols,s1810_colnames)

income_df = load_acs_tables('data/ACS/S1901_IncomeDistribution.csv',s1901_cols,s1901_colnames)

employment_df = load_acs_tables('data/ACS/S2301_EmploymentStatus.csv',s2301_cols,s2301_colnames)

zero_vehicle_df = load_acs_tables('data/ACS/B08201_HHVehicle.csv',b08201_cols,b08201_colnames)
```

```python
hh_cols_to_convert = [
    'total_hh',
    'avg_hh_size',
    'hh_with_ppl_under18',
    'hh_with_ppl_over60',
    'hh_with_ppl_over65',
    'hh_live_alone',
    'hh_live_alone_over65',
    'mobile_homes',
    'owner_occ_units',
    'renter_occ_units',
]

for col in hh_cols_to_convert:
  if col in hh_family_df.columns:
    hh_family_df[col] = pd.to_numeric(hh_family_df[col], errors='coerce')

# Rate of households with children under 18
hh_family_df['rate_hh_under18'] = (
    hh_family_df['hh_with_ppl_under18'] / hh_family_df['total_hh']
) * 100

# Rate of households living alone
hh_family_df['rate_hh_live_alone'] = (
    hh_family_df['hh_live_alone'] / hh_family_df['total_hh']
) * 100

# Rate of seniors living alone 
hh_family_df['rate_hh_alone_over65'] = (
    hh_family_df['hh_live_alone_over65'] / hh_family_df['total_hh']
) * 100

# Rate of mobile homes
hh_family_df['rate_mobile_homes'] = (
    hh_family_df['mobile_homes'] / hh_family_df['total_hh']
) * 100

# Rate of owner-occupied units 
hh_family_df['rate_owner_occ'] = (
    hh_family_df['owner_occ_units'] / hh_family_df['total_hh']
) * 100

# Rate of renter-occupied units 
hh_family_df['rate_renter_occ'] = (
    hh_family_df['renter_occ_units'] / hh_family_df['total_hh']
) * 100


hh_family_df
```

```python
zero_vehicle_cols_to_convert = ['zero_vehicle_hh', 'total_hh_vehicles']

for col in zero_vehicle_cols_to_convert:
  if col in zero_vehicle_df.columns:
    zero_vehicle_df[col] = pd.to_numeric(zero_vehicle_df[col], errors='coerce')


zero_vehicle_df['rate_zero_vehicle'] = (
    zero_vehicle_df['zero_vehicle_hh'] / zero_vehicle_df['total_hh_vehicles']
) * 100
  
zero_vehicle_df
```

```python
disability_cols_to_convert = [
    'total_pop_disability',
    'disabled_count',
    'disabled_under5',
    'disabled_5to17',
    'disabled_65to74',
    'disabled_over75',
    'disability_rate',
]

for col in disability_cols_to_convert:
  if col in disability_df.columns:
    disability_df[col] = pd.to_numeric(disability_df[col], errors='coerce')

disability_df['disabled_under18_count'] = (
    disability_df['disabled_under5'].fillna(0)
    + disability_df['disabled_5to17'].fillna(0)
)
disability_df['disabled_over65_count'] = (
    disability_df['disabled_65to74'].fillna(0)
    + disability_df['disabled_over75'].fillna(0)
)

# Rate of total disabled population who are seniors
disability_df['rate_disabled_over65'] = (
    disability_df['disabled_over65_count'] / disability_df['disabled_count']
) * 100

disability_df['rate_disabled_under18'] = (
    disability_df['disabled_under18_count'] / disability_df['disabled_count']
) * 100

disability_df
```

```python
income_cols_to_convert = [
    'total_hh_income',
    'hh_income_under10k',
    'hh_income_10k_to_15k',
    'hh_income_15k_to_25k',
    'hh_income_25k_to_35k',
    'hh_income_35k_to_50k',
    'hh_income_50k_to_75k',
    'hh_income_75k_to_100k',
    'hh_median_income',
]

for col in income_cols_to_convert:
  if col in income_df.columns:
    income_df[col] = pd.to_numeric(income_df[col], errors='coerce')

income_df['low_income_hh_count'] = (
    income_df['hh_income_under10k'].fillna(0)
    + income_df['hh_income_10k_to_15k'].fillna(0)
    + income_df['hh_income_15k_to_25k'].fillna(0)
    + income_df['hh_income_25k_to_35k'].fillna(0)
)

income_df['rate_low_income_hh'] = (
    income_df['low_income_hh_count'] / income_df['total_hh_income']
) * 100

income_df
```

## Combine ACS and FEMA data

```python
nri_df = pd.read_csv("data/NRI_Table_CensusTracts/NRI_Table_CensusTracts.csv")

nri_df
```

```python
acs_dfs = [
    age_df,
    hh_family_df,
    poverty_df,
    disability_df,
    income_df,
    employment_df,
    zero_vehicle_df,
]

master_acs = acs_dfs[0]

# Loop through the remaining dataframes and merge them one by one
for df in acs_dfs[1:]:
  master_acs = master_acs.merge(df, on='geo_id', how='outer')

print(f"Master ACS shape: {master_acs.shape}")
```

```python
nri_df['TRACTFIPS'] = nri_df['TRACTFIPS'].astype(str).str.zfill(11)

master_df = master_acs.merge(
    nri_df, left_on='geo_id', right_on='TRACTFIPS', how='inner'
)

columns_to_drop = [
    'total_hh_vehicles',
    'total_hh_income',
    'total_pop_disability',
    'total_pop_poverty',
]

master_df = master_df.drop(columns=columns_to_drop, errors='ignore')

print(f'Final Master Dataset Shape: {master_df.shape}')

master_df
```
